# Assessment 2: Grid World Problem Solution

This notebook demonstrates the application of our custom `value_iteration` package to solve the specific Grid World Markov Decision Process (MDP) defined in the assessment brief.

## Problem Formulation
We are evaluating a 2x2 grid environment with the following characteristics:
* **States:** Top-Left (TL), Top-Right (TR), Bottom-Left (BL), and Bottom-Right (BR).
* **Terminal State:** The bottom-right square (BR) is an absorbing terminal state.
* **Actions & Dynamics:** The agent can attempt to move left, right, up, or down. However, the environment is stochastic (e.g., attempting to move right from TL has a 0.9 probability of reaching TR, and a 0.1 probability of slipping into BL).
* **Rewards:** Movement incurs various costs (negative rewards), while reaching the terminal state from different directions yields positive rewards.

The goal is to find the optimal policy $\pi^*$ that maximizes the expected cumulative discounted reward from any starting state.

```python
#Import packages
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import sys
import os

# Ensure the notebook can find the package in the parent directory
sys.path.append(os.path.abspath('..'))

# Import the core solver and the formatting utility
from value_iteration import value_iteration, print_results

# Import the specific problem data generator for Assessment 
from examples.gridworld_generators import get_assessment_grid_world
```

## Execution

We will now initialize the environment and run the value iteration algorithm. Unless otherwise specified by the environment constraints, we use a standard discount factor ($\gamma = 0.9$) and a strict convergence threshold ($\theta = 10^{-6}$).

```python
# Load the assessment data
states, transitions, terminal_states = get_assessment_grid_world()

# Execute the algorithm
V, policy = value_iteration(
    states=states, 
    transitions=transitions, 
    terminal_states=terminal_states, 
    gamma=0.9
)

# Output the results
print_results(V, policy)
```

```python
def visualize_grid_world(V, policy):
    """Visualizes the 2x2 Grid World's values and optimal policy."""
    
    # Map the dictionary values into a 2x2 numpy array
    # Layout: [[TL, TR], 
    #          [BL, BR]]
    v_matrix = np.array([
        [V['TL'], V['TR']],
        [V['BL'], V['BR']]
    ])
    
    # Map the policy actions to a text matrix for the labels
    # We replace None with 'TERM' or a star for the terminal state
    def get_action_label(action):
        if action is None: return "★ TERM ★"
        return f"Move {action}"
        
    policy_matrix = np.array([
        [get_action_label(policy['TL']), get_action_label(policy['TR'])],
        [get_action_label(policy['BL']), get_action_label(policy['BR'])]
    ])
    
    # Create the Heatmap
    plt.figure(figsize=(8, 6))
    
    # Use seaborn to create a colored heatmap. 
    # annot=policy_matrix overlays your optimal actions as text!
    ax = sns.heatmap(v_matrix, annot=policy_matrix, fmt="", cmap="YlGnBu", 
                     cbar_kws={'label': 'State Value (V)'}, annot_kws={"size": 12, "weight": "bold"})
    
    # Format the axes to look like a clean grid
    ax.set_xticklabels(['Left', 'Right'], fontsize=12)
    ax.set_yticklabels(['Top', 'Bottom'], fontsize=12, rotation=0)
    plt.title("Grid World: Optimal Values and Policy", fontsize=16, pad=20)
    
    plt.show()

# Run the visualization!
visualize_grid_world(V, policy)
```

## Conclusion

The output above clearly defines the optimal action to take from any square on the grid. 

Because the algorithm evaluates the expected future rewards mathematically (accounting for the 10% or 20% chances of slipping into adjacent squares), the resulting policy perfectly balances the risk of moving through high-cost areas against the final reward of reaching state BR.
